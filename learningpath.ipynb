### Block 3: Container im Detail: Die Sandbox

  * Schwerpunkt dieses Teils ist die Vermittlung der Tatsache, dass ein Container eigentlich ziemlich trivial ist: Ein Prozess, der auf dem Container-Dateisystem operiert
  * Zu Beginn nutzen wir das hello-world-Image, das ja im Wesentlichen nur ein executable enthält, was der export in ein Archiv beweist
  * Das Kopieren von Dateien in / aus dem Container kann hier bereits demostriert werden
  * Erst durch den Wechsel auf das nginx-Image können wir uns mit exec auf den laufenden Container verbinden und darin einen Shell-Prozess starten
  * ℹ️ Environments sowie die Kopplung mehrere Container wird auf docker compose verschoben
